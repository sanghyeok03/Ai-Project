```python
# 라이브러리 임포트
import pandas as pd
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.decomposition import PCA
from sklearn.cluster import DBSCAN
import matplotlib.pyplot as plt
import pandas.api.types as ptypes

# 1. 데이터 불러오기
try:
    df = pd.read_csv('bank_transactions_data_2.csv') # CSV 파일 로드
    print("Data loaded successfully from bank_transactions_data_2.csv.")
except FileNotFoundError as e:# 파일이 없을 때 에러 출력
    print(f"Error: {e}. Please check the file name and path.")
    exit()
except Exception as e:# 그 외 예외 처리
    print(f"An unexpected error occurred during file loading: {e}")
    exit()

# 2. 사용 컬럼 선택
cols = [
    'TransactionID', 'AccountID', 'TransactionAmount', 'TransactionDate',
    'TransactionType', 'Location', 'DeviceID', 'IP Address', 'MerchantID',
    'Channel', 'CustomerAge', 'CustomerOccupation', 'TransactionDuration',
    'LoginAttempts', 'AccountBalance', 'PreviousTransactionDate'
]

missing_cols = [col for col in cols if col not in df.columns]
if missing_cols:
    print(f"Warning: Missing columns: {missing_cols}")
    cols = [col for col in cols if col in df.columns]
    if not cols:
        raise ValueError("No valid columns found. Exiting.")

df_selected = df[cols].copy() # 선택한 컬럼만 새로운 DataFrame에 복사

# 3. 범주형 변수에 대한 Label Encoding 수행
label_encoders = {}
for col in df_selected.columns:
    if ptypes.is_object_dtype(df_selected[col]) or ptypes.is_categorical_dtype(df_selected[col]):
        le = LabelEncoder()
        df_selected[col] = df_selected[col].fillna("Missing") # 결측값은 'Missing'으로 대체
        df_selected[col] = le.fit_transform(df_selected[col]) # 숫자로 변환
        label_encoders[col] = le # 이후 변환용으로 저장
    else:
        df_selected[col] = pd.to_numeric(df_selected[col], errors='coerce') # 숫자로 변환 시도

# 4. 결측치 제거
df_selected.dropna(inplace=True)
if df_selected.empty:
    print("Error: No data remaining after dropna.")
    exit()

# 5. 정규화
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_selected) # 평균 0, 분산 1로 스케일링

# 6. 학습/검증 분할 및 텐서 변환
X_train, X_val = train_test_split(X_scaled, test_size=0.2, random_state=42)
X_train_tensor = torch.FloatTensor(X_train)
X_val_tensor = torch.FloatTensor(X_val)
train_loader = torch.utils.data.DataLoader(X_train_tensor, batch_size=128, shuffle=True)  # 배치 단위로 학습

# 7. 오토인코더 정의
input_dim = X_scaled.shape[1]

class Autoencoder(nn.Module):
    def __init__(self, input_dim, encoding_dim=10):
        super(Autoencoder, self).__init__()
        encoding_dim = min(max(1, encoding_dim), input_dim) # 인코딩 차원 조정
        self.encoder = nn.Sequential(
            nn.Linear(input_dim, 64),
            nn.ReLU(),
            nn.Linear(64, encoding_dim)
        )
        self.decoder = nn.Sequential(
            nn.Linear(encoding_dim, 64),
            nn.ReLU(),
            nn.Linear(64, input_dim)
        )

    def forward(self, x):
        encoded = self.encoder(x)
        decoded = self.decoder(encoded)
        return decoded

model = Autoencoder(input_dim=input_dim) # 모델 초기화
criterion = nn.MSELoss() # 손실 함수: 평균 제곱 오차
optimizer = optim.Adam(model.parameters(), lr=1e-3) # 옵티마이저 설정

# 8. 모델 학습
num_epochs = 50
for epoch in range(num_epochs):
    model.train()
    train_loss = 0
    for batch in train_loader:
        optimizer.zero_grad()
        outputs = model(batch)
        loss = criterion(outputs, batch) # 입력과 복원 값 간 오차 계산
        loss.backward()
        optimizer.step()
        train_loss += loss.item()
    print(f"Epoch {epoch+1}/{num_epochs}, Loss: {train_loss / len(train_loader):.4f}")

# 9. 인코딩된 데이터 추출
model.eval()
with torch.no_grad():
    encoded_data = model.encoder(torch.FloatTensor(X_scaled)).numpy() # 인코딩 결과 추출

# 10. DBSCAN 클러스터링
if encoded_data.shape[0] >= 5:
    dbscan = DBSCAN(eps=0.5, min_samples=5)
    labels = dbscan.fit_predict(encoded_data) # 클러스터링 수행
    if len(labels) == len(df_selected.index):
        outliers_dbscan = df.loc[df_selected.index[labels == -1]] # 라벨 -1은 이상치
        print(f"DBSCAN 이상치 탐지 수: {len(outliers_dbscan)}")
    else:
        print("DBSCAN 레이블 수와 인덱스 수 불일치.")
else:
    print("DBSCAN을 수행할 샘플 수 부족.")
    labels = np.zeros(encoded_data.shape[0]) # 모든 샘플이 정상

# 11. PCA 시각화
if encoded_data.shape[0] > 1 and encoded_data.shape[1] >= 2:
    reduced = PCA(n_components=2).fit_transform(encoded_data)
    if len(labels) == len(reduced):
        plt.figure(figsize=(8, 6))
        plt.scatter(reduced[:, 0], reduced[:, 1], c=labels, cmap='Spectral', s=5)
        plt.title("DBSCAN 결과 (PCA 2D)")
        plt.xlabel("PC1")
        plt.ylabel("PC2")
        plt.colorbar()
        plt.show()
    else:
        print("PCA 결과와 라벨 수 불일치")

# 12. 재구성 오류 계산
with torch.no_grad():
    X_tensor = torch.FloatTensor(X_scaled)
    reconstructed = model(X_tensor) # 입력 복원
    reconstruction_error = torch.mean((X_tensor - reconstructed) ** 2, dim=1).numpy() # 각 샘플별 MSE 계산

# 13. 재구성 오류 시각화
plt.figure(figsize=(8, 4))
plt.hist(reconstruction_error, bins=50, color='skyblue', edgecolor='black')
plt.title("재구성 오류 분포")
plt.xlabel("MSE 재구성 오류")
plt.ylabel("빈도")
plt.grid(True)
plt.show()

# 14. 이상치 탐지 기준 적용 (예: 상위 5% 재구성 오류 기준)
threshold = np.percentile(reconstruction_error, 95) # 상위 5% 오류 값을 기준으로 이상치 정의
anomaly_mask = reconstruction_error > threshold
outliers_reconstruction = df.loc[df_selected.index[anomaly_mask]] # 이상치만 추출
print(f"재구성 오류 기반 이상치 탐지 수: {len(outliers_reconstruction)}")
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.cluster import DBSCAN
from sklearn.neighbors import NearestNeighbors
import matplotlib.pyplot as plt
import gc # garbage collection (메모리 회수용)

# 1. 데이터 로딩
try:
    df = pd.read_csv('bank_transactions_data_2.csv') # CSV 파일 불러오기
    print("✅ 데이터 로딩 완료. 총 샘플 수:", len(df))
except Exception as e: # 예외 발생 시 에러 메시지 출력
    print(f"❌ 데이터 로딩 오류: {e}")
    raise e # 에러를 다시 발생시켜 종료

# 2. 샘플 수 제한 (RAM 과부하 방지)
if len(df) > 10000:
    df = df.sample(n=10000, random_state=42) # 샘플 10,000개로 제한
    print("✅ 샘플 수 10,000개로 제한")

# 3. 우선 사용하려는 컬럼 정의
priority_cols = [
    'TransactionID', 'AccountID', 'TransactionAmount', 'TransactionDate',
    'TransactionType', 'Location', 'DeviceID', 'IP Address', 'MerchantID',
    'Channel', 'CustomerAge', 'CustomerOccupation', 'TransactionDuration',
    'LoginAttempts', 'AccountBalance', 'PreviousTransactionDate'
]

# 4. Label Encoding + 날짜 처리
df_processed = df.copy() # 원본 보호용 복사
for col in priority_cols:
    if col not in df.columns:
        continue # 존재하지 않는 컬럼은 스킵
    if df[col].dtype == 'object':
      # 문자열(범주형) 컬럼은 결측치 'missing'으로 채우고 정수 인코딩
        df_processed[col] = df[col].fillna('missing')
        df_processed[col] = LabelEncoder().fit_transform(df_processed[col])
    elif 'Date' in col:
      # 날짜 컬럼은 datetime으로 변환 후 최소값 기준 초(second)로 변환
        df_processed[col] = pd.to_datetime(df[col], errors='coerce')
        df_processed[col] = (df_processed[col] - df_processed[col].min()).dt.total_seconds()
    else:
      # 나머지는 가능한 수치형으로 변환
        df_processed[col] = pd.to_numeric(df[col], errors='coerce')

# 5. 수치형 컬럼만 필터링
numeric_df = df_processed[priority_cols].select_dtypes(include=['int64', 'float64', 'int32', 'float32'])
numeric_cols = numeric_df.columns.tolist()

if not numeric_cols:
    raise ValueError("❌ 수치형 컬럼이 없습니다. 확인 필요")

print("✅ 사용되는 수치형 컬럼:", numeric_cols)

# 6. 결측치 제거 + float32 변환
df_filtered = df_processed[numeric_cols].dropna().astype('float32')

# 7. 스케일링
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_filtered)

# 8. k-distance plot (DBSCAN용 eps 선택 참고)
if X_scaled.shape[0] >= 5:
    neigh = NearestNeighbors(n_neighbors=5)
    nbrs = neigh.fit(X_scaled)
    distances, _ = nbrs.kneighbors(X_scaled)
    distances = np.sort(distances[:, 4]) # 5번째 최근접 이웃 거리 기준 정렬

    plt.figure(figsize=(8, 4))
    plt.plot(distances)
    plt.xlabel('Points sorted by distance')
    plt.ylabel('5-NN Distance')
    plt.title('k-distance plot (eps 선택 참고용)')
    plt.grid(True)
    plt.show()
else:
    print("⚠️ 샘플 수 부족으로 k-distance plot 생략")

# 9. DBSCAN 실행
eps_values = [0.5, 1.0] # 다양한 밀도 기준 거리
min_samples_values = [5] # 최소 이웃 수

for eps in eps_values:
    for min_samples in min_samples_values:
        if min_samples > X_scaled.shape[0]: # 샘플 수보다 이웃 수가 많을 수 없음
            print(f"⚠️ Skipping eps={eps}, min_samples={min_samples}: 샘플 수 부족")
            continue

        dbscan = DBSCAN(eps=eps, min_samples=min_samples)
        labels = dbscan.fit_predict(X_scaled) # 클러스터 할당
        n_outliers = np.sum(labels == -1) # 라벨이 -1인 경우는 이상치
        print(f"✅ eps={eps}, min_samples={min_samples} → 이상치 수: {n_outliers} / 총 샘플: {len(labels)}")

# 10. 메모리 회수
gc.collect()
```

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import DBSCAN
from sklearn.metrics import mean_squared_error
from sklearn.neighbors import NearestNeighbors
from keras.models import Model
from keras.layers import Input, Dense
from keras import regularizers
import matplotlib.pyplot as plt
import gc

# 1. 데이터 로딩
df = pd.read_csv("bank_transactions_data_2.csv")
print("✅ 데이터 로드 완료. 샘플 수:", len(df))

# 2. 수치형 컬럼 자동 선택
numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()

# 3. 결측치 없는 컬럼만 필터링
used_cols = [col for col in numeric_cols if df[col].notnull().all()]
print("✅ 사용 가능한 수치형 컬럼:", used_cols)

# 4. 데이터 제한 (RAM 초과 방지)
X = df[used_cols].values
X = X[:10000]  # 최대 10,000개로 제한
X = StandardScaler().fit_transform(X.astype(np.float32))

# 5. k-distance plot 시각화
k = 5
neighbors = NearestNeighbors(n_neighbors=k)
neighbors_fit = neighbors.fit(X)
distances, indices = neighbors_fit.kneighbors(X)
distances = np.sort(distances[:, k-1]) # k번째 이웃 거리 정렬

plt.figure(figsize=(8, 4))
plt.plot(distances)
plt.title(f"{k}-distance Plot (DBSCAN)")
plt.xlabel("Points sorted by distance")
plt.ylabel(f"{k}th Nearest Neighbor Distance")
plt.grid(True)
plt.show()

# 6. DBSCAN으로 이상치 탐지
print("\n📌 DBSCAN 이상치 탐지 결과:")
for eps in [0.5, 1.0, 1.5]:
    for min_samples in [5, 10]:
        dbscan = DBSCAN(eps=eps, min_samples=min_samples)
        labels = dbscan.fit_predict(X)
        n_outliers = np.sum(labels == -1)
        print(f"▶ eps={eps}, min_samples={min_samples} ➜ 이상치 탐지 수: {n_outliers}")

# 7. Autoencoder 모델 정의
encoding_dim = min(10, X.shape[1]) # 인코딩 차원 설정
input_dim = X.shape[1]

input_layer = Input(shape=(input_dim,))
encoded = Dense(encoding_dim, activation='relu',
                activity_regularizer=regularizers.l1(1e-5))(input_layer)
decoded = Dense(input_dim, activation='linear')(encoded)
autoencoder = Model(inputs=input_layer, outputs=decoded)
autoencoder.compile(optimizer='adam', loss='mse')

# 8. Autoencoder 학습
print("\n🧠 Autoencoder 학습 시작...")
autoencoder.fit(X, X, epochs=50, batch_size=64, shuffle=True, verbose=1)

# 9. 재구성 오차 기반 이상치 탐지
reconstructions = autoencoder.predict(X) # 복원된 데이터
mse = np.mean(np.power(X - reconstructions, 2), axis=1) # MSE 계산
threshold = np.percentile(mse, 95)  # 상위 5%를 이상치로 간주
n_anomalies = np.sum(mse > threshold)
print(f"\n✅ Autoencoder 이상치 탐지 수: {n_anomalies} (Threshold: {threshold:.5f})")

# 10. RAM 정리
gc.collect()

# DBSCAN 실행 및 결과 저장
eps_values = [0.5, 1.0]
min_samples_values = [5]

for eps in eps_values:
    for min_samples in min_samples_values:
        if min_samples > X_scaled.shape[0]:
            print(f"⚠️ Skipping eps={eps}, min_samples={min_samples}: 샘플 수 부족")
            continue

        dbscan = DBSCAN(eps=eps, min_samples=min_samples)
        labels = dbscan.fit_predict(X_scaled)
        n_outliers = np.sum(labels == -1)
        print(f"✅ eps={eps}, min_samples={min_samples} → 이상치 수: {n_outliers} / 총 샘플: {len(labels)}")

        # 결과를 원래 DataFrame에 붙이기
        df_result = df_filtered.copy()
        df_result['DBSCAN_label'] = labels

        # 전체 결과 저장
        result_filename = f'dbscan_results_eps{eps}_min{min_samples}.csv'
        df_result.to_csv(result_filename, index=False)

        # 이상치만 따로 저장
        outliers_df = df_result[df_result['DBSCAN_label'] == -1]
        outlier_filename = f'dbscan_outliers_eps{eps}_min{min_samples}.csv'
        outliers_df.to_csv(outlier_filename, index=False)

        print(f"📁 결과 저장 완료 → {result_filename}, {outlier_filename}")
```
